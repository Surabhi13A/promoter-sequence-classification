We see that the class is the label used during supervised learning.  
During training, the model learns the relationship between target variable(y) and input features(X)

The class column is a target variable, representing whether the dna sequence is a promoter(+) or not(-).  
This is stored as a separate variable(y), so the model can learn the mapping from input variables to the correct class

Here we see that identifier is a unique identifer(name) and doesnt contribute towards identifying if the sequence is a promoter or not.  
Since it doesn't contain any property of the DNA it can be dropped. Including it would cause the model to remember the names instead of learning any patterns

And sequence and one_hot_sequence tell us about the same data in differnet formats, and since models we are using here - logistic regression require numerical features, we will be using the on hot encoded column, and not the string representation of the DNA

```python
X = df["one_hot_sequence"]
X.head()
```

```python
X.shape
```

```python
type(X)
```

```python
type(X.iloc[0])
```

```python
X.iloc[0].shape
```

Here we flatten each one-hot encoded DNA sequence from a 57 × 4 matrix into a 228-dimensional feature vector.  

This is necessary because scikit-learn estimators expect the input X to have the shape (n_samples, n_features), where each row is one sample and each column is one feature.  

Our one-hot encoded data has the shape (n_samples, 57, 4), which is a 3D array. By flattening each sequence, we transform the data into the required shape (n_samples, 228).  

```python
X = X.apply(lambda seq: seq.flatten())
X.head()
```

```python
X.shape
```

```python
X[0].shape
```

```python
X.iloc[0].shape
```

```python
X = np.stack(X)
X.shape
```

Now we change the class labels from '+' and '-' to 1 and 0 respectively.  
This is for convention, + and - works just as well

```python
y
```

### Data Splitting

We split the data by 80% and 20%.  
The model learns on the training data, i.e., adjusts its weights and biases and finds a mathematical relation between features and target.  
This model is evaluated on the unseen test data to see how it performs on neew data 
We have startified i.e., the promoter and not promoters proportion in the data is preserved.

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20, shuffle=True, random_state=13, stratify=y)
```

```python
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```
